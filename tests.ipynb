```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.constants import pi

x = np.linspace(0, 50, 100) / 1000
D = 100 / 1000
area = pi / 4 * (D**2 - (D - 2 * x) ** 2)
inercia = pi / 64 * (D**4 - (D - 2 * x) ** 4)
plt.plot(x, area, "r", label="Area (m²)")
plt.plot(x, inercia, "b", label=r"M. inércia (m$^4$)")
plt.xlabel("Espessura (m)")
plt.legend()
```

```python
from src.parametros_py import calcular_deslocamento
import matplotlib.pyplot as plt
import numpy as np

espessura = np.linspace(1, 10, 100) / 1000
deslocamento = []
for val in espessura:
    _temp = calcular_deslocamento(
        densidade=2700, comprimento=1, espessura=val, modulo_elasticidade=68.9e6
    )
    deslocamento.append(_temp)


plt.plot(espessura, deslocamento, "r")
plt.xlabel("Espessura (m)")
plt.ylabel("Deslocamento (m)")
```
